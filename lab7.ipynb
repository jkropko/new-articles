## Problem 1
In this lab, you will use orchestration (using `prefect`) and automation (using GitHub actions) to set up a research literature pipeline for yourself. This pipeline will

1. Access the OpenAlex API. OpenAlex is like Google Scholar, but better because it has more information about individual articles, and it has an API that enables literature review work that uses code instead of point-and-clicking everything.

2. Search for articles that match search terms you supply. That way you can find new research that is highly relevant to you. We will filter this search to articles that only became available in the last week. 

3. Pull all the information needed to create a Zotero entry for each article (stored in an .ris file that can be uploaded directly to Zotero with all the information needed for a bibliographioc). Zotero is a free citation management tool. It is a requirement for anyone doing serious research. If you don't have Zotero on your computer, download it right now: https://www.zotero.org/

4. Create an .ris file: a file that contains all the information needed to create a complete Zotero record for each article, including the full abstracts

5. Download the PDF of each article, when available

6. Store both the .ris file and the PDFs on GitHub, ready to download at your convenience

The hope is that, more than being an exercise, this lab sets up an important research tool for you to use throughout the rest of grad school and in your academic career.
